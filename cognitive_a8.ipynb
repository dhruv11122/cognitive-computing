```python
text = """
Artificial Intelligence is one of my favorite areas of technology because it combines creativity with problem solving.
I am fascinated by how AI systems can learn from data and make useful predictions.
Technologies such as machine learning, computer vision, and natural language processing are changing the way people work.
I especially enjoy learning how machines can understand images, text, and human language.
AI has applications in healthcare, education, finance, entertainment, and many other fields.
I believe Artificial Intelligence will play an important role in shaping the future of technology.
"""

print(text)
```

```python
import re

clean_text = text.lower()
clean_text = re.sub(r'[^\w\s]', '', clean_text)

print(clean_text)
```

```python
import nltk

# Download required tokenizer data
nltk.download('punkt')
nltk.download('punkt_tab')

# Tokenize into words
word_tokens = nltk.word_tokenize(clean_text)

# Tokenize into sentences
sentence_tokens = nltk.sent_tokenize(text)

print("Word Tokens:")
print(word_tokens)

print("\nSentence Tokens:")
print(sentence_tokens)
```

```python
# Using Python's split()
split_tokens = clean_text.split()

# Using NLTK's word_tokenize()
nltk_tokens = nltk.word_tokenize(clean_text)

print("Python split() output:")
print(split_tokens)

print("\nNLTK word_tokenize() output:")
print(nltk_tokens)

print("\nNumber of tokens:")
print("split():", len(split_tokens))
print("word_tokenize():", len(nltk_tokens))
```

```python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

filtered_words = [word for word in word_tokens if word.lower() not in stop_words]

print("Words after removing stopwords:")
print(filtered_words)
```

```python
from nltk.probability import FreqDist

# Calculate frequency distribution
freq_dist = FreqDist(filtered_words)

print("Word Frequency Distribution:")
for word, frequency in freq_dist.items():
    print(word, ":", frequency)

# Plot the frequency distribution
freq_dist.plot(10, title="Word Frequency Distribution")
```

```python
# Extract words containing only alphabets
alphabet_words = re.findall(r'\b[a-zA-Z]+\b', text)

print("Words containing only alphabets:")
print(alphabet_words)
```

```python
# Remove stopwords
q2_filtered_words = [
    word.lower()
    for word in alphabet_words
    if word.lower() not in stop_words
]

print("Words after removing stopwords:")
print(q2_filtered_words)
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in q2_filtered_words]

print("Stemmed words:")
print(stemmed_words)
```

```python
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in q2_filtered_words]

print("Lemmatized words:")
print(lemmatized_words)
```

```python
print("Comparison of Stemming and Lemmatization:\n")

print("Original words:")
print(q2_filtered_words)

print("\nStemmed words:")
print(stemmed_words)

print("\nLemmatized words:")
print(lemmatized_words)

print("\nExplanation:")
print("Stemming removes prefixes or suffixes to produce a shorter root-like form.")
print("Lemmatization converts words to their meaningful dictionary or base form.")
print("Stemming is faster and useful when speed is important.")
print("Lemmatization is preferred when meaningful and linguistically correct words are required.")
```

```python
from sklearn.feature_extraction.text import CountVectorizer

texts = [
    "The laptop has excellent performance and a powerful processor.",
    "The smartphone has a beautiful display and excellent camera.",
    "The laptop has a powerful battery and excellent performance."
]

vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(texts)

print("Vocabulary:")
print(vectorizer.get_feature_names_out())

print("\nBag of Words representation:")
print(bow_matrix.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

print("Vocabulary:")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF scores:")
print(tfidf_matrix.toarray())
```

```python
import numpy as np

feature_names = tfidf_vectorizer.get_feature_names_out()

for i, text_item in enumerate(texts):
    scores = tfidf_matrix[i].toarray().flatten()

    # Get indices of the top 3 TF-IDF scores
    top_indices = np.argsort(scores)[-3:][::-1]

    print(f"Text {i + 1}:")
    for index in top_indices:
        print(feature_names[index], ":", round(scores[index], 3))

    print()
```

```python
import re
import nltk

text_ai = """
Artificial Intelligence enables machines to learn from data and perform intelligent tasks.
AI is widely used in healthcare, education, finance, and automation.
Machine learning helps AI systems identify patterns and make predictions.
AI can improve efficiency and solve complex real-world problems.
"""

text_blockchain = """
Blockchain is a decentralized technology used to store data securely.
It records transactions in a distributed and tamper-resistant ledger.
Blockchain is widely associated with cryptocurrencies and digital assets.
It can also be used for secure payments, supply chains, and digital identity.
"""

# Preprocess and tokenize AI text
ai_clean = re.sub(r'[^\w\s]', '', text_ai.lower())
ai_tokens = nltk.word_tokenize(ai_clean)

# Preprocess and tokenize Blockchain text
blockchain_clean = re.sub(r'[^\w\s]', '', text_blockchain.lower())
blockchain_tokens = nltk.word_tokenize(blockchain_clean)

print("AI tokens:")
print(ai_tokens)

print("\nBlockchain tokens:")
print(blockchain_tokens)
```

```python
# Convert tokens into sets
ai_set = set(ai_tokens)
blockchain_set = set(blockchain_tokens)

# Calculate Jaccard Similarity
intersection = ai_set.intersection(blockchain_set)
union = ai_set.union(blockchain_set)

jaccard_similarity = len(intersection) / len(union)

print("Common words:", intersection)
print("Jaccard Similarity:", round(jaccard_similarity, 4))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Create TF-IDF vectors for both texts
q4_vectorizer = TfidfVectorizer()
q4_tfidf = q4_vectorizer.fit_transform([text_ai, text_blockchain])

# Calculate cosine similarity
cosine_sim = cosine_similarity(q4_tfidf[0:1], q4_tfidf[1:2])

print("Cosine Similarity:", round(cosine_sim[0][0], 4))
```

```python
print("Analysis:")
print("Jaccard Similarity compares the presence or absence of unique words in the two texts.")
print("Cosine Similarity considers the TF-IDF weights of the words, so it gives more importance to informative words.")
print("For these technology texts, Cosine Similarity gives better insights because it considers word importance and frequency.")
print("Jaccard Similarity is simpler and useful when we only want to compare the overlap of unique words.")
```

```python
from textblob import TextBlob

reviews = [
    "The laptop has excellent performance and a beautiful display. I am very happy with its speed.",
    "The phone battery is terrible and the camera quality is disappointing. I regret buying it.",
    "The headphones are decent and work well for everyday use. The sound quality is acceptable."
]

# Find polarity and subjectivity for each review
for i, review in enumerate(reviews):
    analysis = TextBlob(review)

    print(f"Review {i + 1}:")
    print("Polarity:", round(analysis.sentiment.polarity, 2))
    print("Subjectivity:", round(analysis.sentiment.subjectivity, 2))
    print()
```

```python
for i, review in enumerate(reviews):
    polarity = TextBlob(review).sentiment.polarity

    if polarity > 0:
        sentiment = "Positive"
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print(f"Review {i + 1}: {sentiment}")
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Select positive reviews
positive_reviews = []

for review in reviews:
    polarity = TextBlob(review).sentiment.polarity

    if polarity > 0:
        positive_reviews.append(review)

# Combine all positive reviews
positive_text = " ".join(positive_reviews)

# Create word cloud
wordcloud = WordCloud(width=800, height=400, background_color="white").generate(positive_text)

# Display word cloud
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Positive Reviews")
plt.show()
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer

training_text = """
Artificial Intelligence is transforming the way people interact with technology.
Machine learning allows computers to learn patterns from data and make predictions.
Deep learning uses neural networks to solve complex problems in different fields.
Natural language processing helps computers understand and generate human language.
Computer vision enables machines to analyze and interpret images and videos.
AI is being used in healthcare, education, finance, transportation, and entertainment.
As technology continues to improve, intelligent systems are becoming more useful.
Learning about AI helps students understand modern technologies and their applications.
The future of artificial intelligence will depend on responsible development and innovation.
"""

# Create and fit the tokenizer
tokenizer = Tokenizer()
tokenizer.fit_on_texts([training_text])

# Convert text into word sequences
sequences = tokenizer.texts_to_sequences([training_text])

print("Word Index:")
print(tokenizer.word_index)

print("\nTokenized Sequence:")
print(sequences[0])
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

# Create input sequences
input_sequences = []

for line in training_text.split("."):
    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i + 1]
        input_sequences.append(n_gram_sequence)

# Find maximum sequence length
max_sequence_len = max(len(seq) for seq in input_sequences)

# Pad sequences
from tensorflow.keras.preprocessing.sequence import pad_sequences

input_sequences = np.array(
    pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre')
)

# Separate input and target word
X = input_sequences[:, :-1]
y = input_sequences[:, -1]

# Number of words in vocabulary
vocab_size = len(tokenizer.word_index) + 1

# Build LSTM model
model = Sequential([
    Embedding(input_dim=vocab_size, output_dim=50, input_shape=(max_sequence_len - 1,)),
    LSTM(100),
    Dense(vocab_size, activation='softmax')
])

model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.summary()
```

```python
# Train the model
model.fit(X, y, epochs=500, verbose=0)

# Generate text using the most probable next word
def generate_text(seed_text, next_words):
    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences([seed_text])[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding='pre'
        )

        predictions = model.predict(token_list, verbose=0)[0]
        predicted = np.argmax(predictions)

        output_word = ""

        for word, index in tokenizer.word_index.items():
            if index == predicted:
                output_word = word
                break

        if output_word == "":
            break

        seed_text += " " + output_word

    return seed_text


# Generate 2-3 lines of text
generated_text = generate_text("Artificial", 18)

print("Generated Text:")

for seed in ["Artificial", "Machine", "Technology"]:
    print(generate_text(seed, 8))
```
