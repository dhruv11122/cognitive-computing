```python
import pandas as pd

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

personalized_entries = [
    {
        "question": "how can i update my billing details",
        "answer": "You can update billing details from the billing section.",
        "keywords": "billing update invoice payment",
        "category": "billing"
    },
    {
        "question": "how do i change my registered mobile number",
        "answer": "You can change your registered mobile number from account settings.",
        "keywords": "account mobile update profile",
        "category": "account"
    }
]

faq_entries = fixed_entries + personalized_entries

df = pd.DataFrame(faq_entries)

print(df)
```

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for index, row in df.iterrows():
        keyword_words = set(row["keywords"].lower().split())

        score = len(query_words.intersection(keyword_words))

        if score > 0:
            results.append(
                {
                    "question": row["question"],
                    "answer": row["answer"],
                    "confidence_score": score
                }
            )

    results = sorted(
        results,
        key=lambda x: x["confidence_score"],
        reverse=True
    )

    return results


query = input("Enter your query: ")

print(score_query(query, df))
```

```python
def same_category(category_name, df):
    return df[df["category"] == category_name]["question"]


print(same_category("billing", df))
```

```python
def score_query_with_tie(query, df):

    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        keyword_words = set(row["keywords"].lower().split())

        score = len(query_words.intersection(keyword_words))

        results.append(
            {
                "question": row["question"],
                "answer": row["answer"],
                "score": score
            }
        )

    max_score = max(result["score"] for result in results)

    print("Highest matching entries:")

    for result in results:
        if result["score"] == max_score:
            print(result)


# Query producing tie
score_query_with_tie("fee", df)


# Query without tie
score_query_with_tie("password", df)
```
