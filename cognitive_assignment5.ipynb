```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original Array:", arr)

print("After adding 2:", arr + 2)

print("After multiplying by 3:", arr * 3)

print("After dividing by 2:", arr / 2)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])

reverse_arr = arr[::-1]

print("Original Array:", arr)
print("Reversed Array:", reverse_arr)
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])

values, counts = np.unique(x, return_counts=True)

max_value = values[np.argmax(counts)]

print("Most frequent value:", max_value)
print("Indices:", np.where(x == max_value)[0])
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

max_value = values[np.argmax(counts)]

print("Most frequent value:", max_value)
print("Indices:", np.where(y == max_value)[0])
```

```python
arr = np.array([[10,20,30],
                [40,50,60],
                [70,80,90]])

print("1st row, 2nd column:", arr[0,1])

print("3rd row, 1st column:", arr[2,0])
```

```python
ucs420_dhruv = np.linspace(10,100,25)

print("Array:")
print(ucs420_dhruv)

print("Dimension:", ucs420_dhruv.ndim)

print("Shape:", ucs420_dhruv.shape)

print("Total Elements:", ucs420_dhruv.size)

print("Data Type:", ucs420_dhruv.dtype)

print("Total Bytes:", ucs420_dhruv.nbytes)
```

```python
transpose_arr = ucs420_dhruv.reshape(5,5)

print("Transpose using reshape:")
print(transpose_arr)
```

```python
print("Using T attribute:")
print(ucs420_dhruv.T)
```

```python
ucs420_dhruv = np.array([[10,20,30,40],
                         [50,60,70,80],
                         [90,15,20,35]])

print("Original Array:")
print(ucs420_dhruv)

print("Mean:", np.mean(ucs420_dhruv))

print("Median:", np.median(ucs420_dhruv))

print("Maximum:", np.max(ucs420_dhruv))

print("Minimum:", np.min(ucs420_dhruv))

print("Unique Elements:", np.unique(ucs420_dhruv))
```

```python
reshaped_ucs420_dhruv = ucs420_dhruv.reshape(4,3)

print("Reshaped Array:")
print(reshaped_ucs420_dhruv)
```

```python
resized_ucs420_dhruv = np.resize(ucs420_dhruv,(2,3))

print("Resized Array:")
print(resized_ucs420_dhruv)
```
