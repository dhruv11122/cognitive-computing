```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

new_temp = temperature + 2

print("Original temperatures:", temperature)
print("Corrected temperatures:", new_temp)
```

```python
fahrenheit = (9/5) * new_temp + 32

print("Temperature in Fahrenheit:", fahrenheit)
```

```python
above_32 = new_temp[new_temp > 32]

print("Readings greater than 32°C:", above_32)
```

```python
count = np.sum(new_temp > 32)

print("Number of readings above 32°C:", count)
```

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print(steps)
```

```python
print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))
```

```python
day_total = np.sum(steps, axis=0)

print("Total steps for each day:", day_total)
```

```python
user_total = np.sum(steps, axis=1)

print("Total steps for each user:", user_total)
```

```python
max_pos = np.unravel_index(np.argmax(steps), steps.shape)

print("Position of maximum steps:", max_pos)
print("Maximum steps:", steps[max_pos])
```

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])
```

```python
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
```

```python
XT = X.T

print("X transpose:")
print(XT)
```

```python
product = X.T @ X

print("X.T @ X:")
print(product)
```

```python
inverse = np.linalg.inv(product)

print("Inverse:")
print(inverse)
```

```python
beta = inverse @ X.T @ y

print("Coefficients:")
print(beta)
```

```python
new_user = np.array([5, 40, 7])

prediction = new_user @ beta

print("Predicted assistance score:", prediction)
```
